Dataset: Tsanas, A. & Xifara, A. (2012). Energy Efficiency [Dataset]. UCI Machine Learning Repository. https://doi.org/10.24432/C51307.

```python
import pandas as pd
import numpy as np 
from sklearn.svm import SVR
from sklearn.ensemble import RandomForestRegressor
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import PolynomialFeatures, StandardScaler
from sklearn.metrics import r2_score, mean_squared_error
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv("energy_data.csv")
```

```python
df.head()
```

```python
df.isna().sum().sum()
```

```python
columnNames = ["Relative Compactness", "Surface Area", "Wall Area", "Roof Area", "Overall Height", "Orientation", "Glazing Area", "Glazing Area Distribution", "Heating Load", "Cooling Load"]
df.columns = columnNames
df.head()
```

```python
df.info()
```

```python
X = df.iloc[:,:-2]
y1 = df.iloc[:,-2]
y2 = df.iloc[:,-1]

X_train, X_test, y_train1, y_test1, y_train2, y_test2 = train_test_split(X, y1, y2, test_size=0.3, random_state=42)
scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.fit_transform(X_test)
```

```python
models = {
    "Linear Regression": LinearRegression(),
    "Polynomial Regression": make_pipeline(PolynomialFeatures(degree=2), LinearRegression()),
    "Random Forest": RandomForestRegressor(n_estimators=100),
    "SVR": SVR(C=100)
}
```

```python
for name, model in models.items():
    model.fit(X_train, y_train1)
    y_pred = model.predict(X_test)
    mse = mean_squared_error(y_test1, y_pred)
    r2 = r2_score(y_test1, y_pred)
    print(f"model: {name}, mean squared error: {mse:.4f}, R²: {r2:.4f}")
```

Ranking:
1. SVR -> lowest MSE, highest R^2
2. Linear Regression 
3. Random Forest
4. Polynomial Regression -> overfitting

```python
for name, model in models.items():
    model.fit(X_train, y_train2)
    y_pred = model.predict(X_test)
    mse = mean_squared_error(y_test2, y_pred)
    r2 = r2_score(y_test2, y_pred)
    print(f"model: {name}, mean squared error: {mse:.4f}, R²: {r2:.4f}")
```

Ranking:
1. SVR -> lowest MSE, highest R²
2. Random Forest
3. Linear Regression
4. Polynomial Regression -> overfitting
